## Part 1 — What machine learning is

**1. What kind of problem is machine learning the right tool for?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Problems where you can <b>recognise</b> the right answer but cannot <b>state the rule</b> that produces it.</li>
        <li>The difficulty is one of <b>specification</b>, not of computation: spam is easy to spot and hard to describe.</li>
        <li>Instead of writing the rule, you supply examples and let a procedure search for a rule consistent with them.</li>
        <li>If you can state the rule, write it directly. A learned model is harder to verify, harder to maintain, and needs data.</li>
    </ul>
    </p>
</details>

**2. Give three situations in which you should <i>not</i> use machine learning.**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li><b>The rule is known</b> — write it. Nobody should train a model to test whether a number is even.</li>
        <li><b>The data is not representative</b> of where the model will be deployed; its measured accuracy then says little about real performance.</li>
        <li><b>Errors are catastrophic and unexplainable</b>, with no human reviewing the output.</li>
        <li><b>The data encodes an injustice</b> you would be automating, now with an appearance of objectivity.</li>
        <li><b>A simple baseline already suffices</b> — often a threshold on one variable solves most of the problem and needs no maintenance.</li>
    </ul>
    </p>
</details>

**3. Distinguish the expected risk from the empirical risk, and state why the distinction matters. <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li><b>Expected risk</b> <code>R(f) = E[L(f(x), y)]</code> is the average loss over the true data distribution — including data that does not exist yet. It is what we care about.</li>
        <li><b>Empirical risk</b> <code>R̂(f) = (1/n) Σ L(f(xᵢ), yᵢ)</code> is the average loss on the finite sample we hold. It is what we can compute.</li>
        <li>The expected risk is <b>not computable</b>, because the distribution is unknown. So we minimise the empirical risk as a proxy.</li>
        <li>The two differ, and a flexible enough model can drive the empirical risk to zero by <b>memorising the sample</b> while the expected risk stays terrible. That gap is overfitting.</li>
        <li>Everything about held-out data, validation and cross-validation exists to <b>estimate the quantity we cannot compute</b>.</li>
    </ul>
    </p>
</details>

**4. Why can a model that achieves zero error on its training data still be useless?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Zero training error is achievable by <b>memorising</b> rather than by finding a pattern — storing every example and reciting its answer.</li>
        <li>Performance on data used for fitting is <b>always optimistic</b> and sometimes meaningless as an estimate of future performance.</li>
        <li>The only informative question is how the model behaves on data <b>it has never seen</b>, which is why a test set is held out.</li>
    </ul>
    </p>
</details>

## Part 2 — Kinds of learning

**5. On what basis does the supervised / unsupervised / self-supervised taxonomy divide methods?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>On <b>where the target comes from</b>, not on the algorithms — the same estimator can serve more than one setting.</li>
        <li><b>Supervised</b>: a person produced the labels.</li>
        <li><b>Unsupervised</b>: there is no target at all.</li>
        <li><b>Self-supervised</b>: the target is manufactured from the input itself, by hiding part of it.</li>
    </ul>
    </p>
</details>

**6. What practically limits most supervised learning projects, and why?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The <b>labels</b> — not algorithms, not compute.</li>
        <li>Someone must produce them, correctly, in quantity, and for many problems that is expensive or impossible.</li>
        <li>This is precisely why unsupervised and self-supervised methods dominate wherever data is abundant and annotation is not.</li>
    </ul>
    </p>
</details>

**7. In notebook 02 the self-supervised section runs the same estimator as the supervised regression. So what actually differs? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Mechanically, nothing — same <code>LinearRegression</code>, same call.</li>
        <li>What differs is <b>the provenance of the target</b>: it came from the data itself rather than from an annotator.</li>
        <li>That makes the supervision <b>free</b>, which is what allows the technique to scale to unannotated text and images.</li>
        <li>The pretext task is not the goal: nobody wants a flavanoid predictor. Solving it forces the model to represent how the inputs relate, and that representation transfers.</li>
    </ul>
    </p>
</details>

**8. Why is it harder to tell whether an unsupervised method has succeeded?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>There is <b>no ground truth</b> to compare against, so no accuracy to report.</li>
        <li>A clustering algorithm will <b>always</b> return clusters, whether or not the data contains groups.</li>
        <li>Whether the structure found is meaningful is a <b>judgement about the domain</b>, and cannot be delegated to a metric.</li>
    </ul>
    </p>
</details>

## Part 3 — The workflow

**9. Why must the split come before scaling, feature selection and inspection?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Because the test set only answers the question 'how does this behave on unseen data' <b>while it remains unseen</b>.</li>
        <li>Any step that <b>learns</b> from data — a mean for scaling, a set of categories, a ranking of features — must learn from training data only.</li>
        <li>Otherwise information from the test set reaches the model and the estimate becomes optimistic. That is <b>data leakage</b>.</li>
        <li>It happens without any error appearing in the code: every line runs as intended and the resulting number is meaningless.</li>
    </ul>
    </p>
</details>

**10. What is a baseline for, and what does its absence cost you?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>It is the <b>simplest thing that could work</b>: predict the majority class, predict the mean, threshold one variable.</li>
        <li>Without it a score has no scale. 95% accuracy may be excellent, or worse than answering 'no' every time.</li>
        <li>On imbalanced data the trivial baseline is often <b>unbeatable on accuracy</b> and useless in practice — which is itself the lesson.</li>
    </ul>
    </p>
</details>

**11. Why does a <code>Pipeline</code> matter more than tidiness?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Preprocessing steps <b>learn parameters</b> (a mean, a standard deviation, a vocabulary), so they are part of the model, not preparation of the data.</li>
        <li>Inside a pipeline they are refitted on the training portion <b>every time</b> the pipeline is fitted, including within each cross-validation fold.</li>
        <li>Done by hand, the statistics are eventually computed over the whole dataset — leakage, silently, with no failure to notice.</li>
    </ul>
    </p>
</details>

**12. You are asked to choose an evaluation metric. When should you choose it, and why then? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li><b>Before modelling</b>, as part of framing the problem.</li>
        <li>The right metric follows from which error is more harmful, which is a question about <b>consequences</b>, not about mathematics.</li>
        <li>A metric chosen after seeing results is a metric chosen to <b>flatter</b> them: with enough candidates, one will look good by chance.</li>
        <li>Committing in advance is what makes the eventual number a test rather than a search.</li>
    </ul>
    </p>
</details>

## Part 4 — How models mislead

**13. Notebook 03 reaches 77% accuracy on labels assigned by a coin flip. Explain how. <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The 20 'best' features were selected using <b>the whole dataset</b>, before splitting.</li>
        <li>With 5000 random features, some correlate with the labels <b>by chance</b> — including the labels of rows that later become the test set.</li>
        <li>So the selection encoded information about the test labels, and the model inherited it. The test set was consulted and is no longer unseen.</li>
        <li>Selection feels harmless because it fits no model, but it <b>learns from data</b> and therefore belongs inside the training fold.</li>
        <li>More noise features make the illusion <b>stronger</b>, because there are more opportunities for a chance correlation.</li>
    </ul>
    </p>
</details>

**14. A screening model reports 99% accuracy on a problem with 1% positives. What do you ask next?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>What does the <b>trivial model</b> score? Answering 'negative' always gives 99% here.</li>
        <li>What is the <b>recall on the positive class</b> — the fraction of cases of interest actually caught?</li>
        <li>Accuracy is dominated by the majority class, which is the class nobody built the system for.</li>
    </ul>
    </p>
</details>

**15. What is a shortcut feature, and why can no metric detect one?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>A column that is a <b>consequence of the label</b> rather than a predictor of it — a treatment code, a scheduled follow-up.</li>
        <li>It boosts every score, including cross-validated ones, because the information genuinely is in the training data.</li>
        <li>It fails at deployment, where the value <b>does not exist yet</b> at the moment a prediction is needed.</li>
        <li>Detecting it requires knowing <b>when and why each field was recorded</b>. That is a question about the data-generating process, and no metric asks it.</li>
    </ul>
    </p>
</details>

**16. Why is a single train/test split a weak way to report a result?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>It is <b>one draw from a distribution</b>, reported as though it were a property of the model.</li>
        <li>In notebook 03 the spread across seeds was 0.083 — larger than most claimed improvements between competing methods.</li>
        <li>Anyone can choose the seed that flatters them, and a reader cannot tell from a single number.</li>
        <li>Report the <b>mean and the spread</b>, or the result is incomplete.</li>
    </ul>
    </p>
</details>

## Part 5 — History and limits

**17. What pattern recurs in the history of AI, and what should you take from it?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>A genuine advance, followed by <b>claims outrunning evidence</b>, followed by a correction severe enough to cut funding — the perceptron in 1969, neural networks in the early 1990s.</li>
        <li>Many advances came from <b>data and compute</b> rather than new principles: AlexNet's architecture was recognisably LeNet from 1989.</li>
        <li>The useful habit is distinguishing a <b>demonstrated result</b> from an <b>extrapolated promise</b>. Rosenblatt's convergence theorem was true; 'it will reproduce itself' was not.</li>
        <li>Ask what changed. A scaling result and a conceptual one license very different predictions.</li>
    </ul>
    </p>
</details>

**18. Minsky and Papert showed a perceptron cannot represent XOR. Was that a fair reason to abandon neural networks? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The mathematics was <b>correct</b>: no single linear boundary separates XOR.</li>
        <li>But it applies to a <b>single layer</b>. Multi-layer networks were already known to be more expressive; what was missing was a way to train them.</li>
        <li>That arrived with backpropagation in 1986, and a hidden layer disposes of XOR in a few lines — as you will see in Lesson 9.</li>
        <li>The lesson is about <b>scope</b>: a correct result about a restricted case was received as a verdict on the whole approach, and cost roughly fifteen years.</li>
    </ul>
    </p>
</details>

**19. In what sense is a trained model not objective?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>It is a <b>compressed summary of its training data</b>, including whatever bias produced that data.</li>
        <li>If past decisions discriminated, a model fitted to them reproduces the discrimination — with an <b>appearance of neutrality</b> that makes it harder to contest than a human decision.</li>
        <li>These methods find <b>correlation</b>; nothing in empirical risk minimisation separates a cause from a coincidence that predicts well in this sample.</li>
        <li>So a model can be highly accurate and completely wrong about why — and fail abruptly when conditions change.</li>
    </ul>
    </p>
</details>

**20. Beyond accuracy, what properties belong in the framing of a problem?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Whether a decision can be <b>explained</b> to the person it affects.</li>
        <li>Whether errors are <b>recoverable</b>, and whether anyone notices when the model fails.</li>
        <li><b>Who bears the cost</b> of a wrong answer — often not the person deploying the model.</li>
        <li>These are engineering requirements, not an afterword. They belong in step 1 of the workflow, where they can still change what you build.</li>
    </ul>
    </p>
</details>
