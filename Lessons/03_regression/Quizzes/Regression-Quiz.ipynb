## Part 1 — The model and its cost

**1. What exactly does a linear model claim about the world?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>That each feature contributes a <b>fixed amount per unit</b>, and that the contributions <b>add up</b>.</li>
        <li>Concretely: every square metre is worth the same 2,400 euros whether it is the fortieth or the two-hundredth.</li>
        <li>It is a strong claim and often false — the hundred-and-first square metre is probably worth less than the fiftieth.</li>
        <li>The response is not to abandon it but to <b>know what you assumed</b>, and to check the residuals for structure it missed.</li>
    </ul>
    </p>
</details>

**2. Why not simply add up the errors to measure fit?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Because they cancel. A prediction 50,000 too high and one 50,000 too low sum to zero and look perfect.</li>
        <li>Every error must count as a <b>positive amount</b>, which is the requirement squaring meets.</li>
        <li>Absolute value also meets it — the next question is why we prefer squares.</li>
    </ul>
    </p>
</details>

**3. Give three reasons for squared error rather than absolute error.**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>It is <b>differentiable everywhere</b>; |x| has a corner at zero that complicates optimisation.</li>
        <li>It <b>punishes one large error more</b> than several small ones, which matches most intuitions about a bad prediction.</li>
        <li>Under Gaussian noise it <b>is</b> maximum likelihood estimation — the deep reason, and what makes it more than a convention.</li>
        <li>The cost: squared error is <b>sensitive to outliers</b>, because one absurd value contributes its error squared.</li>
    </ul>
    </p>
</details>

**4. Where does the ½ in the cost function come from, and does it change the answer?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Differentiating a square brings down a factor of 2; the half cancels it, leaving a clean gradient.</li>
        <li>It does <b>not</b> move the minimum: scaling a function by a positive constant leaves its minimiser unchanged.</li>
        <li>Pure convenience, and worth recognising so it does not look like something meaningful.</li>
    </ul>
    </p>
</details>

## Part 2 — The exact solution

**5. Explain the normal equation as a picture, without algebra. <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Each feature is a <b>direction</b> the model can move in; everything it can predict lies on the flat surface those directions span.</li>
        <li>The true targets are a point that almost never lies on that surface.</li>
        <li>The best possible fit is the <b>closest point on the surface</b> — the shadow, or perpendicular projection.</li>
        <li>Why perpendicular? If the leftover error had a component <b>along</b> the surface, you could move that way and do better. So at the optimum the error is at right angles to every feature — which is what the equations say, and where the name <i>normal</i> comes from.</li>
    </ul>
    </p>
</details>

**6. How do we know the solution to the normal equation is a minimum rather than a maximum or saddle?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The second derivative is <code>(1/m) Xᵀ X</code>, which is <b>positive semi-definite</b>: for any v, <code>vᵀXᵀXv = ‖Xv‖² ≥ 0</code>.</li>
        <li>So the cost is <b>convex</b>, and any stationary point is a global minimum.</li>
        <li>This is a genuinely useful property that most later methods lack — <b>there are no local minima to get stuck in</b>. Lesson 9's networks are full of them.</li>
    </ul>
    </p>
</details>

**7. Name the three situations where the exact solution is unusable, and say which is the dangerous one.**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li><b>Exactly redundant columns</b> — the same measurement twice, so <code>XᵀX</code> is singular and has no inverse.</li>
        <li><b>Nearly redundant columns</b> — the inverse exists but is enormous, and small data changes produce large coefficient changes.</li>
        <li><b>Size</b> — inverting an n×n matrix costs about n³ operations, out of reach at large n.</li>
        <li>The <b>second</b> is the dangerous one, because nothing visibly fails. Notebook 3 shows a coefficient moving from 9,261 to 45,307 across four splits of one dataset, with the matrix invertible every time.</li>
    </ul>
    </p>
</details>

## Part 3 — Gradient descent

**8. Describe gradient descent without notation.**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>You are on a hillside in fog: you cannot see the valley, but you can feel which way the ground slopes.</li>
        <li>Take a step downhill, feel again, repeat. The gradient is the direction of steepest ascent, so we walk <b>against</b> it.</li>
        <li>The learning rate is the length of the stride, and the danger is equally intuitive: <b>stride too far and you cross the valley</b> and end up higher on the far slope.</li>
    </ul>
    </p>
</details>

**9. The gradient with respect to w_j contains a factor of x_j. What does that mean in practice? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Each example pulls the coefficient in proportion to <b>two</b> things: how wrong the prediction was, <b>and</b> how large that feature was for that example.</li>
        <li>So a 200 m² house that is badly mispriced moves the area coefficient far more than a 40 m² one with the same error.</li>
        <li>The gradient is a <b>weighted vote</b> and the weights are the feature values — which is exactly why features on very different scales cause trouble.</li>
        <li>The intercept gradient has no x in it, so every example gets an equal vote there. That is why the intercept crawls while a large-valued feature races.</li>
    </ul>
    </p>
</details>

**10. What happens if the learning rate is too large, and where does the threshold come from?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The step overshoots the minimum and lands higher on the opposite side; the next step overshoots further. The cost <b>increases</b>, often to <code>nan</code> within a few dozen iterations.</li>
        <li>For a quadratic bowl of curvature c, convergence requires <b>α &lt; 2/c</b> (Lesson 2).</li>
        <li>With several features the binding constraint is the <b>largest</b> curvature — the largest feature variance. So the safe learning rate is set by your worst-scaled feature.</li>
    </ul>
    </p>
</details>

## Part 4 — Flexibility and overfitting

**11. A linear model is fitted to a curved relationship. How, and why is it still called linear?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>By adding powers of the feature as extra columns: <code>ŷ = w₁t + w₂t² + b</code>.</li>
        <li>It is linear <b>in the coefficients</b>, which is all the derivation ever required — the inputs may be any transformation you can compute.</li>
        <li>Same normal equation, same code, one more column. The trick extends to interactions and logarithms.</li>
    </ul>
    </p>
</details>

**12. In notebook 2, training error falls at every degree while test error bottoms out at degree 3. What does that tell you about model selection?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>That <b>training error cannot be used to choose a model</b>: it falls monotonically with flexibility and never signals that anything is wrong.</li>
        <li>The degree-12 model fits the training data best — measured by root mean squared error (RMSE), the square root of the mean squared error — and is <b>ten times worse</b> on unseen data — test RMSE 182 against 16.9.</li>
        <li>Selecting by training error would have picked the worst model, with the training error congratulating you the whole way.</li>
        <li>This is Lesson 1's empirical-versus-expected risk argument with numbers in it.</li>
    </ul>
    </p>
</details>

**13. Mechanically, what goes wrong inside an overfitted polynomial? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Its coefficients become enormous — 247,514 against 411 for the well-behaved fit — and they work in <b>opposition</b>.</li>
        <li>With 21 points and 12 coefficients many arrangements pass close to every point; the one least squares finds has terms that nearly cancel, one pushing up where the next pushes down.</li>
        <li>The cancellation fails exactly where a training point sits, and <b>between the points nothing constrains the swing</b>.</li>
        <li>That observation is what regularisation acts on: if huge opposing coefficients are the symptom, charge for size.</li>
    </ul>
    </p>
</details>

## Part 5 — Regularisation

**14. Why must features be scaled before Ridge or Lasso, and is this the same reason as in Lesson 2?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The penalty is a <b>sum over coefficients</b>, so the same feature in metres and in kilometres attracts penalties differing by a factor of a thousand.</li>
        <li>An unscaled feature is therefore penalised according to its <b>units</b> rather than its importance.</li>
        <li>It is <b>not</b> the same reason as Lesson 2. There, scaling was about optimisation speed — the condition number. Here it is about <b>correctness</b>: without it the penalty means something arbitrary.</li>
    </ul>
    </p>
</details>

**15. Why does Lasso produce exact zeros while Ridge does not? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Think of the penalty as a <b>budget</b>: the best fit you can afford is where the error contours first touch the affordable region.</li>
        <li>Charging squares makes that region a <b>disc</b>; charging absolute values makes it a <b>diamond</b> standing on its corners.</li>
        <li>A disc is smooth, so contact happens at a generic point with both coefficients non-zero.</li>
        <li>A diamond has <b>corners, and its corners lie on the axes</b> — a corner is a point where one coefficient is exactly zero. Corners protrude, so an expanding contour touches one first.</li>
        <li>So the sparsity is geometric, not a numerical tolerance.</li>
    </ul>
    </p>
</details>

**16. Ridge adds λI to the matrix being inverted. Why does that fix the singularity?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Adding a positive constant to the diagonal shifts <b>every eigenvalue up by λ</b>, so none of them can be zero.</li>
        <li><code>XᵀX + λI</code> is therefore invertible for any λ &gt; 0, even when <code>XᵀX</code> is not.</li>
        <li>Intuitively: least squares is <b>flat</b> in the direction where two columns duplicate each other, so it has no unique lowest point. The penalty <b>tilts the floor</b>, and a single lowest point appears.</li>
    </ul>
    </p>
</details>

**17. In the ridge table, training error gets worse as λ grows while test error first improves. What is being traded?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Fit on the data we have, in exchange for fit on data we do not.</li>
        <li>At λ = 0.01 the largest coefficient falls from <b>3,097,038,010 to 365</b> and test RMSE from <b>24,656 to 23</b>, while training RMSE rises only from 13.6 to 18.0. A touch of rigidity, and three orders of magnitude of test error.</li>
        <li>The trade <b>stops paying</b>: at λ = 100, test RMSE is 228, worse than no penalty at all, because the model is now too rigid to follow the curve.</li>
        <li>Both ends are failures. Finding the middle honestly needs cross-validation — Lesson 5.</li>
        <li>Handout Section 4.4 adds a second thing the penalty buys, which the table does not show: at λ = 0.01 the condition number of the matrix being inverted falls from 1.8 × 10¹⁹ to 23,100, which is what makes the number reproducible across machines.</li>
    </ul>
    </p>
</details>

**18. Lasso keeps only <code>area_sqm</code> at λ = 40,000. Has it discovered that area is the most important feature?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Not in any absolute sense. It kept the feature that buys the <b>most accuracy per unit of coefficient</b> under that particular budget.</li>
        <li>The order of elimination depends on <b>scaling</b> and on <b>λ</b>, both of which are choices.</li>
        <li>So "Lasso selected these features" is never a complete statement without saying at which λ and on which scaling.</li>
        <li>It is a useful tool for producing a shortlist, not an objective ranking of importance.</li>
    </ul>
    </p>
</details>

## Part 6 — Reading coefficients

**19. What does a coefficient actually mean, and what is the clause everybody forgets?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The change in the prediction for a one-unit change in that feature, <b>with every other feature held fixed</b>.</li>
        <li>That clause does enormous work. The coefficient on bedrooms is not "what a bedroom is worth" but what a bedroom is worth <b>to a house whose floor area does not change</b> — a strange object.</li>
        <li>It is an <b>association</b>, not a causal effect: nothing in the fit distinguishes a cause from a coincidence that predicts well in this sample.</li>
    </ul>
    </p>
</details>

**20. In notebook 1 the coefficient on distance is recovered within 0.7% and the one on bedrooms is out by 16.6%. Why the difference? <b>(reasoning)</b>**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>Correlation with the other features. Distance correlates with area at 0.00; bedrooms at <b>0.77</b>.</li>
        <li>Correlated features <b>compete to explain the same variation</b>, so the split between them is unstable and the individual estimates wobble.</li>
        <li>The rule: <b>a coefficient is only as trustworthy as its feature is independent</b>.</li>
        <li>And on a real dataset the same wobble would be there with nothing to reveal it — you would simply have a number, and no way to know which coefficients to believe.</li>
    </ul>
    </p>
</details>

**21. With two columns carrying the same fact, least squares gave 9,261 and 45,307 on different splits while predictions stayed accurate. What is and is not broken?**

<details>
<summary>
    <font size='3', color='darkgreen'><b>Answer</b></font>
</summary>
    <p>
    <ul>
        <li>The <b>predictions</b> are fine: the combined effect is stable at about 2,420 against a true 2,400.</li>
        <li>The <b>explanation</b> is noise. The two coefficients swing in opposite directions to compensate for each other.</li>
        <li>Anyone reading them to learn what drives prices would draw nonsense, confidently.</li>
        <li>Ridge splits the effect evenly and stays stable across splits, because two moderate coefficients cost less under a squared penalty than one huge positive and one huge negative. That is what makes coefficients <b>readable</b> when features overlap.</li>
    </ul>
    </p>
</details>
